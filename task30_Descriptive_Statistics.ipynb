```python

from sklearn.datasets import load_iris
import pandas as pd
import numpy as np

# Load Iris dataset
iris = load_iris()

# Convert into a Pandas DataFrame
df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

# Display first five rows
print(df.head())
```

```python
print(df.shape)
```

```python
print(df.columns)
```

```python
data = df.select_dtypes(include='number')

print(data.describe())
```

```python
print("Mean:\n", data.mean())
print("Median:\n", data.median())
print("Mode:\n", data.mode())
print("Range:\n", data.max() - data.min())
print("Variance:\n", data.var())
print("Standard Deviation:\n", data.std())
print("Quartiles:\n", data.quantile([0.25, 0.50, 0.75]))
print("Percentiles:\n", data.quantile([0.10, 0.25, 0.50, 0.75, 0.90]))
```
